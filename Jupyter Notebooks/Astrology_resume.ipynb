```python
def get_zodiac(month, day):
    if month == 1:
        return 'Sagittarius' if day < 20 else 'Capricorn'
    elif month == 2:
        return 'Capricorn' if day < 16 else 'Aquarius'
    elif month == 3:
        return 'Aquarius' if day < 11 else 'Pisces'
    elif month == 4:
        return 'Pisces' if day < 18 else 'Aries'
    elif month == 5:
        return 'Aries' if day < 13 else 'Taurus'
    elif month == 6:
        return 'Taurus' if day < 21 else 'Gemini'
    elif month == 7:
        return 'Gemini' if day < 20 else 'Cancer'
    elif month == 8:
        return 'Cancer' if day < 10 else 'Leo'
    elif month == 9:
        return 'Leo' if day < 16 else 'Virgo'
    elif month == 10:
        return 'Virgo' if day < 30 else 'Libra'
    elif month == 11:
        if day < 23:
            return 'Libra'
        elif day < 29:
            return 'Scorpius'
        else:
            return 'Ophiuchus'
    elif month == 12:
        return 'Ophiuchus' if day < 17 else 'Sagittarius'
    else:
        return 'Invalid month'
```

```python
birthmonth = int(input("what is your birth-month? (1-12): "))
birthday = int(input("what is your birthday?: "))
print('your zodiac is',get_zodiac(birthmonth,birthday))

########
# a program that inputs a user's birthday and gives their precession-adjusted zodiac sign
# input the month and day, output the sign
# chain of elifs to determine sign in order from 1-12 (1 being january and 12 being december)
# there are 13 zodiac signs: capricorn, aquarius, pisces, aries, taurus, gemini, 
# cancer, leo, virgo, libra, scorpio, ophiuchus, and sagittarius
########

# This is approximate because of leap-days
# I may become deranged enough to fine-tune it more so you can know down to the year
```
